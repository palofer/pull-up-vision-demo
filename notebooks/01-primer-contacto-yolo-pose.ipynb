```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path("../src").resolve()))
from main import ejecutar
```

```python
video_path = Path("../data/raw/dominadas-video-optimizado.mp4").resolve()
if not video_path.is_file():
    raise FileNotFoundError(video_path)
print("Vídeo:", video_path)
```

```python
datos, angulos, repeticiones, rutas = ejecutar(video_path)
print("Fotogramas analizados:", len(datos))
print("Repeticiones candidatas:", len(repeticiones))
print("Resultados:", rutas)
```

```python
primer_frame = next(
    (frame for frame in datos if frame["xy"].shape[0] > 0),
    None,
)
if primer_frame is None:
    print("No se detectó ninguna persona en el vídeo")
else:
    print("Frame con persona:", primer_frame["frame"])
    print("Coordenadas:", primer_frame["xy"].shape)
    print("Confianza de keypoints:", primer_frame["conf"].shape)
    print("Confianza de persona:", primer_frame["person_conf"])
```

```python
nombres = {
    5: "hombro izq.", 6: "hombro der.",
    7: "codo izq.", 8: "codo der.",
    9: "muñeca izq.", 10: "muñeca der.",
}

if primer_frame is not None:
    for indice, nombre in nombres.items():
        xy = primer_frame["xy"][0, indice].tolist()
        confianza = primer_frame["conf"][0, indice].item()
        print(f"{nombre}: xy={xy}, confianza={confianza:.2f}")
```
