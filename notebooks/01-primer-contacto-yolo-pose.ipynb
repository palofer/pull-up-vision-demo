```python
import numpy as np
import cv2
from ultralytics import YOLO
from pathlib import Path
```

```python
img = cv2.imread("../data/raw/dominada-vista-trasera-optimizada.jpg")
img.shape
```

```python
weights = "yolo11n-pose.pt"
model = YOLO(weights)
results = model(img)

model_tag = Path(weights).stem
output_dir = Path("../data/processed") / model_tag
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / "dominada-vista-trasera-optimizada.jpg"

saved = cv2.imwrite(str(output_path), results[0].plot())
print(f"Guardado: {saved} — {output_path}")
results[0].plot() 
```

```python
pose = results[0].keypoints

print("Coordenadas:", pose.xy.shape)
print("Confianza de keypoints:", pose.conf.shape)
print("Confianza de la persona:", results[0].boxes.conf)
```

```python
nombres = {
    5: "hombro izq.", 6: "hombro der.",
    7: "codo izq.", 8: "codo der.",
    9: "muñeca izq.", 10: "muñeca der.",
}


for indice, nombre in nombres.items():
    xy = pose.xy[0, indice].tolist()
    confianza = pose.conf[0, indice].item()
    print(f"{nombre}: xy={xy}, confianza={confianza:.2f}")
```
